# Phase 4 — Quantitative Risk Engine Validation Notebook

This notebook demonstrates deterministic evaluation of financial ratios, credit metrics, risk scoring, exact additive contributions, and sensitivity analysis using production `backend.risk_engine` modules.

```python
from backend.data_engine.contracts import SyntheticCompanyConfig
from backend.data_engine.ingest.synthetic import generate_company
from backend.risk_engine.engine import QuantitativeRiskEngine

# 1. Generate canonical synthetic company (seed 1001)
config = SyntheticCompanyConfig(seed=1001, periods=24)
dataset = generate_company(config)
print(f"Company: {dataset.profile.name}, Sector: {dataset.profile.sector}")

# 2. Evaluate quantitative risk
engine = QuantitativeRiskEngine()
report = engine.evaluate_company(dataset)

print(f"Composite Score: {report.composite.score:.2f} ({report.composite.severity})")
print(f"Sensitivity Range: [{report.sensitivity.min_score:.2f}, {report.sensitivity.max_score:.2f}]")
print(f"Rank Stability Score: {report.sensitivity.rank_stability_score:.4f}")
```

```python
# 3. Display dimension score and contribution breakdown
for dim_id, dim_res in report.composite.dimensions.items():
    score_str = f"{dim_res.score:.2f}" if dim_res.score is not None else "N/A"
    print(f"Dimension: {dim_res.name:<32} | Score: {score_str:>6} | Contribution: {dim_res.contribution:6.2f}")
```
